```python
import pandas as pd

df = pd.read_csv(r"C:\Users\Chetan\Downloads\archive\synthetic_ev_battery_5year_50cols_multimfg_3M.csv")

print("FIRST 5 RECORDS")
print(df.head())

print("\nLAST 5 RECORDS")
print(df.tail())

print("\nCOLUMN NAMES")
print(df.columns)

# Select numerical columns
numeric_columns = df.select_dtypes(include='number').columns

print("\nSTATISTICAL ANALYSIS")

for col in numeric_columns:
    data = df[col].dropna()

    
    print("ATTRIBUTE:", col)
  

    print("Frequency:")
    print(data.value_counts().head(10))

    print("Mean:", data.mean())
    print("Mode:", data.mode().iloc[0])
    print("Median:", data.median())
    print("Standard Deviation:", data.std())
    print("Variance:", data.var())
    
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1
    
    print("Interquartile Range:", IQR)
```

```python
# Select one numerical attribute
numeric_columns = df.select_dtypes(include='number').columns
attribute = numeric_columns[0]

# Take first 20 valid values manually
sample_data = []

for value in df[attribute]:
    if pd.notna(value):
        sample_data.append(float(value))
    
    if len(sample_data) == 20:
        break

print("ATTRIBUTE USED:", attribute)
print("SAMPLE DATA:", sample_data)

n = len(sample_data)

# Mean
total = 0

for value in sample_data:
    total = total + value

mean = total / n

# Sort data manually using Bubble Sort
sorted_data = sample_data.copy()

for i in range(n - 1):
    for j in range(n - i - 1):
        if sorted_data[j] > sorted_data[j + 1]:
            temp = sorted_data[j]
            sorted_data[j] = sorted_data[j + 1]
            sorted_data[j + 1] = temp

# Median
if n % 2 == 0:
    median = (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2
else:
    median = sorted_data[n // 2]

# Mode
frequency = {}

for value in sample_data:
    if value in frequency:
        frequency[value] = frequency[value] + 1
    else:
        frequency[value] = 1

max_frequency = 0
mode = sorted_data[0]

for value in frequency:
    if frequency[value] > max_frequency:
        max_frequency = frequency[value]
        mode = value

# Variance
sum_squared_difference = 0

for value in sample_data:
    difference = value - mean
    sum_squared_difference = sum_squared_difference + difference * difference

variance = sum_squared_difference / (n - 1)

# Standard Deviation
standard_deviation = variance ** 0.5

# Q1
if n % 2 == 0:
    lower_half = sorted_data[:n // 2]
else:
    lower_half = sorted_data[:n // 2]

m = len(lower_half)

if m % 2 == 0:
    Q1 = (lower_half[m // 2 - 1] + lower_half[m // 2]) / 2
else:
    Q1 = lower_half[m // 2]

# Q3
if n % 2 == 0:
    upper_half = sorted_data[n // 2:]
else:
    upper_half = sorted_data[n // 2 + 1:]

m = len(upper_half)

if m % 2 == 0:
    Q3 = (upper_half[m // 2 - 1] + upper_half[m // 2]) / 2
else:
    Q3 = upper_half[m // 2]

IQR = Q3 - Q1

print("\nRESULTS")
print("--------------------------------")
print("Mean =", mean)
print("Mode =", mode)
print("Median =", median)
print("Variance =", variance)
print("Standard Deviation =", standard_deviation)
print("Q1 =", Q1)
print("Q3 =", Q3)
print("Interquartile Range =", IQR)
```

```python
# Find categorical and numerical columns
categorical_columns = df.select_dtypes(exclude='number').columns
numeric_columns = df.select_dtypes(include='number').columns

group_column = categorical_columns[0]
value_column = numeric_columns[0]

print("GROUPING ATTRIBUTE:", group_column)
print("NUMERICAL ATTRIBUTE:", value_column)

print("\nGROUPED DATA ANALYSIS")


grouped = df.groupby(group_column)[value_column]

# Frequency
print("\nFREQUENCY")
print(grouped.count())

# Mean
print("\nMEAN")
print(grouped.mean())

# Mode
print("\nMODE")

for group, values in grouped:
    mode_values = values.dropna().mode()
    
    if len(mode_values) > 0:
        print(group, ":", mode_values.iloc[0])
    else:
        print(group, ": No mode")

# Median
print("\nMEDIAN")
print(grouped.median())

# Standard Deviation
print("\nSTANDARD DEVIATION")
print(grouped.std())

# Variance
print("\nVARIANCE")
print(grouped.var())

# Interquartile Range
print("\nINTERQUARTILE RANGE")

for group, values in grouped:
    values = values.dropna()
    
    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)
    IQR = Q3 - Q1
    
    print(group, ":", IQR)
```
