```python
## Perform in detail data analysis of this dataset
import pandas as pd
import numpy as np

# Load CSV file
df = pd.read_csv(r"C:\Users\Chetan\Downloads\archive\synthetic_ev_battery_5year_50cols_multimfg_3M.csv")

print("SHAPE OF DATASET")
print("Rows and Columns:", df.shape)

print("FIRST 5 RECORDS")
print(df.head())

print("LAST 5 RECORDS")
print(df.tail())

print("COLUMN NAMES")
print(df.columns)

```

```python
# Get information and description of dataset.

print("DATA TYPES")
print(df.dtypes)

print("INFO")
print(df.info)

print("DESCRIPTION")
print(df.describe)

```

```python
# 	See if any null values are present. 
## Display count of null values.

print("NULL VALUES IN EACH COLUMN")
print(df.isnull().sum())

print("TOTAL NULL VALUES")
print(df.isnull().sum().sum())
```

```python
# Choose appropriate technique to handle missing values.(imputation with use of inplace)

df["nominal_capacity_Ah"].fillna(df["nominal_capacity_Ah"].mean(),inplace=True)
```

```python
# Use sorting of data in dataframe to display topmost 5 or 8 records based  on one or more column values(conditional filtering)


print("TOP 5 BATTERY CELLS BY SAFETY MARGIN")

top5_safety = df.sort_values(by="safety_margin_index",ascending=False).head(5)

print(top5_safety)

print("TOP 8 RECORDS BY ENERGY DENSITY")

top8_energy = df.sort_values(by="energy_density_Wh_per_kg", ascending=False).head(8)

print(top8_energy)
```

```python
# Get frequency listing of any one relavant column(2 cases)

print("FREQUENCY OF CHEMISTRY TYPES")

print(df["chemistry_type"].value_counts())
```

```python
# Sorting of rows and columns,( implicit and explicit indexing)

print("ROWS SORTED BY NOMINAL CAPACITY")

sorted_rows = df.sort_values(
    by="nominal_capacity_Ah",
    ascending=True
)

print(sorted_rows.head())

# SORTING OF COLUMNS


print("COLUMNS SORTED ALPHABETICALLY")

sorted_columns = df.sort_index(axis=1)

print(sorted_columns.head())

# First 5 rows and first 3 columns
print(df.iloc[0:5, 0:3])

# First 5 rows and selected columns
print(
    df.loc[
        0:4,
        ["cell_id", "chemistry_type", "safety_margin_index"]
    ]
)
```

```python
# Accessing particular row based on certain condition and displaying only one or few columns from it.(3 cases with compound conditions)


print("NMC OR LFP BATTERIES WITH HIGH CAPACITY")

case3 = df.loc[
    (
        (df["chemistry_type"] == "NMC") |
        (df["chemistry_type"] == "LFP")
    ) &
    (df["nominal_capacity_Ah"] >= 3.2),
    ["cell_id", "chemistry_type", "nominal_capacity_Ah"]
]

print(case3)
```

```python
# Minimum and maximum values related analysis

print("MINIMUM AND MAXIMUM VALUES")

print(
    "Minimum Nominal Capacity:",
    df["nominal_capacity_Ah"].min()
)

print(
    "Maximum Nominal Capacity:",
    df["nominal_capacity_Ah"].max()
)

print(
    "Minimum Energy Density:",
    df["energy_density_Wh_per_kg"].min()
)

print(
    "Maximum Energy Density:",
    df["energy_density_Wh_per_kg"].max()
)
```

```python
# Use of group by on one or more columns(2 cases)

print("GROUPBY CASE 2")

group2 = df.groupby(
    ["chemistry_type", "manufacturer"]
)["nominal_capacity_Ah"].mean()

print(group2)
```

```python
# Adding new column to existing dataframe and polulating same using existing columns data.


df["theoretical_energy_Wh"] = (
    df["nominal_capacity_Ah"] *
    df["nominal_voltage_V"]
)

print("NEW COLUMN: THEORETICAL ENERGY")

print(
    df[
        [
            "nominal_capacity_Ah",
            "nominal_voltage_V",
            "theoretical_energy_Wh"
        ]
    ].head()
)
```

```python
# Use of appropriate aggregate functions with groupby.(2 cases)


print("AGGREGATION CASE 1")

aggregate1 = df.groupby("chemistry_type").agg(
    {
        "nominal_capacity_Ah": ["mean", "min", "max"]
    }
)

print(aggregate1)
```

```python
#  Selection on particular groups based on name or condition

print("GROUPS WITH AVERAGE SAFETY MARGIN > 0.5")

chemistry_safety = df.groupby(
    "chemistry_type"
)["safety_margin_index"].mean()

selected_groups = chemistry_safety[
    chemistry_safety > 0.5
]

print(selected_groups)
```

```python
# Find correlation between any two columns values.

print("GROUPS WITH AVERAGE SAFETY MARGIN > 0.8")

chemistry_safety = df.groupby(
    "chemistry_type"
)["safety_margin_index"].mean()

selected_groups = chemistry_safety[
    chemistry_safety > 0.8
]

print(selected_groups)
```

```python
# Find correlation between any two columns values.

print("CORRELATION ANALYSIS")

correlation = df[
    "nominal_capacity_Ah"
].corr(
    df["theoretical_energy_Wh"]
)

print(
    "Correlation between Nominal Capacity and "
    "Theoretical Energy:",
    correlation
)
```

```python
print("MIN-MAX NORMALIZATION")

min_value = df["energy_density_Wh_per_kg"].min()
max_value = df["energy_density_Wh_per_kg"].max()

df["normalized_energy_density"] = (
    (df["energy_density_Wh_per_kg"] - min_value)
    / (max_value - min_value)
)

print(
    df[
        ["energy_density_Wh_per_kg", "normalized_energy_density"]
    ].head()
)
```

```python

# JOINING, MERGING AND CONCATENATION


df1 = df[
    ["cell_id", "manufacturer", "chemistry_type"]
].drop_duplicates("cell_id").head(10).copy()

df2 = df[
    ["cell_id", "nominal_capacity_Ah", "nominal_voltage_V"]
].drop_duplicates("cell_id").head(10).copy()

# Joining
joined_df = df1.join(
    df2.set_index("cell_id"),
    on="cell_id",
    lsuffix="_left",
    rsuffix="_right"
)

print(joined_df)

# MERGING


merged_df = pd.merge(
    df1,
    df2,
    on="cell_id",
    how="inner"
)

print(merged_df)


# CONCATENATION


part1 = df.head(5)
part2 = df.tail(5)

concatenated_df = pd.concat(
    [part1, part2],
    axis=0
)

print(concatenated_df)
```
